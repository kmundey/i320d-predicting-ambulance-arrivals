### Extract GRIB Data

```python
import pandas as pd
import xarray as xr
```

```python
def GRIB_to_df(filepath):
    """
    Converts a GRIB file to a CSV file by reading each message separately.
    """
    import cfgrib
    
    # Get all messages in the file
    messages = cfgrib.open_datasets(filepath)
    
    dfs = []
    for ds in messages:
        # Average across latitude and longitude
        regional_means = ds.mean(dim=["latitude", "longitude"])
        df = regional_means.to_dataframe()
        dfs.append(df)
    
    # Merge all dataframes
    df_merged = pd.concat(dfs, axis=1)

    # drop columns
    df_merged.drop(columns=['valid_time', 'surface', 'step', 'number'], inplace=True)

    # rename columns
    # Rename columns
    df_merged.rename(columns={
        't2m': 'temp_2m',
        'tp': 'total_precipitation',
        'sf': 'snowfall',
        'sd': 'snow_depth',
        'avg_tprate': 'avg_total_precipitation_rate',
        'avg_tsrwe': 'avg_total_snowfall_rate',
        'ptype': 'precipitation_type'
    }, inplace=True)
    
    return df_merged
```

```python
# extract data
era5_2023_filepath = "/Users/kater/Downloads/ERA5_2023_weather_data.grib"
era5_2024_filepath = "/Users/kater/Downloads/ERA5_2024_weather_data.grib"

# Convert data from GRIB to pandas df
era5_2023_df = GRIB_to_df(era5_2023_filepath)
era5_2024_df = GRIB_to_df(era5_2024_filepath)
```

```python
# Merge dataframes into one
df = pd.concat([era5_2023_df, era5_2024_df], axis=0)
```

```python
# Clean up the index datetime format

# --- Step 1: Create a new, clean index ---
def extract_datetime(element):
    
    if isinstance(element, tuple):
        # element[0] is the base timestamp (2024-12-31 18:00:00)
        # element[1] is the time delta (0 days 03:00:00)
        # Summing them gives the correct time (2024-12-31 21:00:00)
        return element[0] + element[1]
    else:
        # Assumes the element is already the correct datetime object
        return element

# Apply the function to the existing index to create the new clean index
clean_index = df.index.map(extract_datetime)

# --- Step 2: Assign the clean index back to the DataFrame ---
df.index = clean_index

# --- Step 3: Ensure the index is a DatetimeIndex and not 'object' ---
# This is crucial after manipulating the index
df.index = pd.to_datetime(df.index)

print(df.info())
```
